# **Data Visualization**

# 1. Histogram - *Univariate*
> We used a histogram to identify the passenger-count ranges in our dataset and to determine how many observations (records) are present in each range.



```python
plt.figure(figsize=(10, 6))

# Create Histogram
plt.hist(
    df["Passenger_Count"],
    bins=20,
)

plt.title("Distribution of Passenger Count")
plt.xlabel("Passenger Count")
plt.ylabel("Number of Observations")

plt.show()
```

# 2. Scatter Plot - *Bivariate*
> The scatter plot shows the relationship between rainfall and average train delay. It indicates that train delays generally increase as rainfall increases, especially during heavy rain conditions

```python
plt.figure(figsize=(12, 6))

#Create Scatter Plot
sns.scatterplot(
    x='Rainfall_mm',
    y='Average_Delay_Min',
    data = df,
    hue='Weather',
    s =60
)

plt.title(
    "Realtionship between Rainfall ana Average Train Delay",
    fontsize=18,
    fontweight='bold',
    pad=20,
)

plt.xlabel('Rainfall (mm)')
plt.ylabel("Average Delay (Minutes)")

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

# 3. Box Plot - *Univariate*
> The box plot shows the distribution of average train delay. It highlights the median, spread of the data, and identifies any outliers in the dataset.

```python
plt.figure(figsize=(8,5))

# Create Box Plot
sns.boxplot(y=df['Average_Delay_Min'])

plt.title(
    'Distribution of Average Train Delay',
    fontsize=14,
    fontweight='bold',
    pad=15,
  )

plt.ylabel('Average Delay (Minutes)')

plt.show()

# Quartiles
Q1 = df['Average_Delay_Min'].quantile(0.25)
Q2 = df['Average_Delay_Min'].median()
Q3 = df['Average_Delay_Min'].quantile(0.75)

# IQR
IQR = Q3 - Q1

# Whiskers
Lower_Whisker = df[df['Average_Delay_Min'] >= (Q1 - 1.5 * IQR)]['Average_Delay_Min'].min()
Upper_Whisker = df[df['Average_Delay_Min'] <= (Q3 + 1.5 * IQR)]['Average_Delay_Min'].max()

print("Lower Whisker:", Lower_Whisker)
print("Q1:", Q1)
print("Median (Q2):", Q2)
print("Q3:", Q3)
print("Upper Whisker:", Upper_Whisker)
```

# 4. Area Chart - *Bivariate*
> This Area Chart compares the total passenger traffic across different railway lines. It helps identify which railway line handles the highest passenger demand, supporting better train planning and resource allocation.

```python
area_df = df.groupby('Line', as_index=False)['Passenger_Count'].sum()

fig = px.area(
    area_df,
    x='Line',
    y='Passenger_Count',
    title='Passenger Traffic Across Railway Lines'
)

fig.update_traces(
    hovertemplate='Line: %{x}<br>Passenger Count: %{y:,.0f}<extra></extra>'
)
fig.show()
```

# 5. Pie Chart - *Univariate*
> We used a pie chart to show the distribution and proportion of different train types in our dataset and understand the share of each train type.

```python
train_type_counts = df['Train_Type'].value_counts()

plt.figure(figsize=(8, 8))

# Create Pie Chart
plt.pie(
    train_type_counts.values,
    labels=train_type_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops= {'fontsize': 14}
)

plt.title('Train Type Distribution', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

# 6. Bar Chart - *Bivariate*
> We used a Bar Chart to compare the passenger count across different railway stations or railway lines and identify which stations or lines have higher or lower passenger counts.

**a) Centeral Line**

```python
central_data = (
    df[df['Line'] == 'Central']
    .groupby('Station')['Passenger_Count']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(central_data.index, central_data.values)

plt.title('Average Passenger Count by Station - Central Line')
plt.xlabel('Station')
plt.ylabel('Average Passenger Count')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**b) Western Line**

```python
western_data = (
    df[df['Line'] == 'Western']
    .groupby('Station')['Passenger_Count']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(western_data.index, western_data.values)

plt.title('Average Passenger Count by Station - Western Line')
plt.xlabel('Station')
plt.ylabel('Average Passenger Count')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**c) Harbour Line**

```python
harbour_data = (
    df[df['Line'] == 'Harbour']
    .groupby('Station')['Passenger_Count']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(harbour_data.index, harbour_data.values)

plt.title('Average Passenger Count by Station - Harbour Line')
plt.xlabel('Station')
plt.ylabel('Average Passenger Count')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**d) Overall Mumbai**

```python
line_avg = (
    df.groupby('Line')['Passenger_Count']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 6))

bars = plt.bar(line_avg.index, line_avg.values)

plt.title('Average Passenger Count by Railway Line')
plt.xlabel('Railway Line')
plt.ylabel('Average Passenger Count')

# Display exact average value
for bar, value in zip(bars, line_avg.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{value:,.0f}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

# 7. Donut Chart - *Univariate*
>This donut chart shows the distribution of different railway station types in the network. It helps understand the proportion of Regular, Junction, and Terminal stations, supporting infrastructure planning and resource allocation.

```python
donut_df = df['Station_Type'].value_counts().reset_index()
donut_df.columns = ['Station_Type', 'Count']

fig = px.pie(
    donut_df,
    names='Station_Type',
    values='Count',
    hole=0.5,
    title='Distribution of Railway Station Types'
)

fig.update_layout(title_x=0.5)

fig.update_traces(
    textfont_size=16,
    textfont_color='white'
)

fig.show()
```

# 8. Line Chart - *Bivariate*

> This line chart shows the variation in average passenger count across different time periods in the Mumbai Local Transit dataset.





```python
time_order = ['Morning', 'Afternoon', 'Evening']

time_avg = (
    df.groupby('Time_Period')['Passenger_Count']
    .mean()
    .reindex(time_order)
)

plt.figure(figsize=(10, 6))

plt.plot(
    time_avg.index,
    time_avg.values,
    marker='o',
    linewidth=2
)

for i, value in enumerate(time_avg.values):
    plt.annotate(
        f'{value:,.0f}',
        (i, value),
        xytext=(0, 10),
        textcoords='offset points',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

plt.title(
    'Average Passenger Count by Time Period',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Time Period')
plt.ylabel('Average Passenger Count')

plt.ylim(2500, 5200)

plt.tight_layout()
plt.show()
```

# 9. Count Plot - *Univariate*
> The Count Plot visualizes the count of each weather condition in the dataset. It helps identify the most and least frequent weather conditions.

```python
plt.figure(figsize=(11,5), dpi=150)

# Create Count Plot
ax = sns.countplot(
    y='Weather',
    data=df,
    hue='Weather',
    palette='Set2',
    legend=False
)

# Add Count Labels on bars
for container in ax.containers:
  ax.bar_label(container, fmt='%d', padding=5, fontsize=10)

plt.title(
    "Distribution of Weather Conditions",
    fontsize=18,
    fontweight='bold',
    pad=20,
)

plt.xlabel("Count", fontsize=12)
plt.ylabel("Weather Condfition", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

# 10. Grouped Bar Chart / Clustered Bar Chart - *Bivariate*
> The grouped bar chart compares different delay reasons across Local, Fast, and AC Local trains. It helps identify which delay reasons are more common for each train type by displaying separate bars for every train category under each delay reason.

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x='Delay_Reason',
    hue='Train_Type',
    palette='Set2'
)

plt.title('Delay Reasons Across Different Train Types', fontsize=14, fontweight='bold')
plt.xlabel('Delay Reason', fontsize=12, labelpad=10)
plt.ylabel('Number of Train Records', fontsize=12)

plt.legend(
    title='Train Type',
    edgecolor='black'
)

plt.tight_layout()
plt.show()
```

# 11. Distplot (Distribution Plot) - *Univariate*
> This Distplot helps us understand the overall distribution of train delays. Most trains have relatively small delays, but a few trains experience significantly higher delays.

```python
# Create a distribution plot for average train delay
sns.histplot(
    df['Average_Delay_Min'],
    kde=True
)

plt.title('Distribution of Average Train Delay', fontsize=14, fontweight='bold')
plt.xlabel('Average Delay (Minutes)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

# 12. Stacked Bar Plot - *Bivariate*
>This stacked bar chart shows passenger distribution across different train types during each time period.

```python
# Calculate total passengers by time period and train type
stacked_data = df.groupby(
    ['Time_Period', 'Train_Type']
)['Passenger_Count'].sum().unstack()

# Set the time period order
time_order = ['Morning', 'Afternoon', 'Evening']
stacked_data = stacked_data.reindex(time_order)

# Create stacked bar chart
ax = stacked_data.plot(
    kind='bar',
    stacked=True,
    figsize=(8, 5)
)

# Remove scientific notation from the Y-axis
ax = plt.gca()
ax.yaxis.get_offset_text().set_visible(False)

# Add exact values inside each section
for container in ax.containers:
    ax.bar_label(
        container,
        labels=[f'{int(bar.get_height()):,}' if bar.get_height() > 0 else ''
                for bar in container],
        label_type='center',
        fontsize=10
    )

plt.title('Passenger Distribution by Time Period and Train Type', fontsize=14, fontweight='bold')
plt.xlabel('Time Period')
plt.ylabel('Total Passenger Count')

# Explain the Y-axis scale
plt.text(
    1.02, 0.5,
    '1 = 1 Million passengers',
    transform=ax.transAxes,
    rotation=90,
    va='center'
)

plt.xticks(rotation=0)
plt.legend(title='Train Type')
plt.tight_layout()
plt.show()
```

# 13. Violin Plot - *Bivariate*
> Violin Plot helps us understand the distribution, concentration, spread, and skewness of train delays across different train types.

```python
plt.figure(figsize=(8,5))

# Create Violin Plot
sns.violinplot(
    data=df,
    x='Train_Type',
    y='Average_Delay_Min'
)

plt.title('Distribution of Average Delay by Train Type', fontsize=13, fontweight='bold')
plt.xlabel('Train Type')
plt.ylabel('Average Delay (Minutes)')

plt.show()
```

# 14. Swarm Plot - *Multivariate*
> This swarm plot helps identify which delay reasons cause higher train delays. It also compares the delay patterns for different delay reasons.

```python
plt.figure(figsize=(10,6))

# Create Swarm Plot
sns.swarmplot(
    data=df,
    x='Delay_Reason',
    y='Average_Delay_Min',
    hue='Delay_Reason',
    palette='Set2',
    size=2.5,
    legend=False
)

plt.title('Average Delay by Delay Reason', fontsize=16, fontweight='bold')
plt.xlabel('Delay Reason', labelpad=10, fontsize=13)
plt.ylabel('Average Delay (Minutes)', fontsize=13)


plt.show()
```
